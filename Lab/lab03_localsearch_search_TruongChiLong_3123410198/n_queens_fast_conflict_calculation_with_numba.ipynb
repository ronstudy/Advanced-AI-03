# Speeding up the objective function calculation using numba

The objective function is evaluated over and over again and it makes sense to spent some time to make it faster. Here we use the just-in-time compilation option from `numba`.


## The n-Queens Problem

* __Goal:__ Find an arrangement of $n$ queens on a $n \times n$ chess board so that no queen is on the same row, column or diagonal as any other queen.

* __State space:__ An arrangement of the queens on the board. We restrict the state space to arrangements where there is only a single queen per column. We represent a state as an integer vector $\mathbf{q} = \{q_1, q_2, \dots, q_n\}$, each number representing the row positions of the queens from left to right. We will call a state a "board."

* __Objective function:__ The number of pairwise conflicts (i.e., two queens in the same row/column/diagonal).
The optimization problem is to find the optimal arrangement $\mathbf{q}^*$ of $n$ queens on the board can be written as:

  > minimize: $\mathrm{conflicts}(\mathbf{q})$
  >
  > subject to: $\mathbf{q} \ \text{contains only one queen per column}$

  Note: the constraint (subject to) is enforced by the definition of the state space.

* __Local improvement move:__ Move one queen to a different row in its column.

* __Termination:__ For this problem there is always an arrangement $\mathbf{q}^*$ with $\mathrm{conflicts}(\mathbf{q}^*) = 0$, however, the local improvement moves might end up in a local minimum. 

## Objective function calculation

### Naive $O(n^2)$ Implementation
Naive implementation with nested loops. This is $O(n^2)$.

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import colors
import math
```

```python
def conflicts_naive(board):
    """Calculate the number of conflicts, i.e., the objective function."""
    
    board = np.array(board)
    
    n = len(board)
    val = 0

    # check horizontal (we do not check vertical since the state space is restricted to one queen per col)
    for i in range(n): val += math.comb(np.sum(board == i), 2)
    #print(f"Horizontal value: {val}")
    
    # check for each queen diagonally up and down (only to the right side of the queen)
    for j in range(n):
        q_up = board[j]
        q_down = board[j]
     
        for jj in range(j+1, n):
            q_up -= 1
            q_down += 1
            if board[jj] == q_up: val += 1
            if board[jj] == q_down: val += 1
        #print(f"Value after queen {j}: {val}")
       
    return(val)    
```

### $O(n)$ Implementation

A better implementation that should be $O(n)$. Idea: 

1. Count how many queens are in each row (no need to check columns) and how many are in each diagonal.
2. Convert the number of $n$ queens in a row (diagonal) into the numbner of pariwise conflicts in $n$ choose 2. We use a faster special case for the `comb()` function to calcualte the conflicts given the number of queens.   

```python
def comb2(n): return n*(n-1)//2 # this is n choose 2 equivalent to math.comb(n, 2); // is int division

def conflicts(board):
    """Calculate the number of conflicts, i.e., the objective function."""

    n = len(board)
    
    horizontal_cnt = [0] * n
    # we have double the diagonals compared to columns!
    diagonal1_cnt = [0] * 2 * n
    diagonal2_cnt = [0] * 2 * n
    
    for i in range(n):
        horizontal_cnt[board[i]] += 1
        diagonal1_cnt[i + board[i]] += 1
        diagonal2_cnt[i - board[i] + n] += 1
    
    return sum(map(comb2, horizontal_cnt + diagonal1_cnt + diagonal2_cnt))
```

### Compiled $O(n)$ Implementation
We can use just-in-time compilation for the conflicts function. Code provided by Troy McNitt.

```python
try:
    from numba import njit
except ModuleNotFoundError:
    import subprocess
    import sys
    subprocess.check_call([sys.executable, "-m", "pip", "install", "numba"])
    from numba import njit

# with jit (cannot use some Python functions like sum and map)
@njit()
def comb2_jit(n): return n*(n-1)//2 # this is n choose 2 equivalent to math.comb(n, 2); // is int division

@njit()
def conflicts_jit(board):
    """Calculate the number of conflicts, i.e., the objective function."""

    n = len(board)
 
    cnt = [0] * (5 * n)
 
    for i in range(n):
        cnt[board[i]] += 1
        cnt[(n) + (i + board[i])] += 1
        cnt[(3 * n) + (i - board[i] + n)] += 1
 
    conflicts = 0
    for x in cnt:
        conflicts += comb2_jit(x)
 
    return conflicts
```

# Compare Timing

Compare runtime on 50 random boards of different sizes.

```python
np.random.seed(1234)
def random_board(n):
    """Creates a random board of size n x n. Note that only a single queen is placed in each column!"""
    
    return(np.random.randint(0,n, size = n))
```

```python
import timeit
 
# run it once to force compilation
conflicts_jit(random_board(4))
    
reps = 20
x = []
naive_times = []
linear_times = []
jit_times = []
 
 
for i in [8] + list(range(50, 500, 50)):
    # do reps replications on random boards
    boards = [random_board(i) for _ in range(reps)]
 
    #print(f"Doing: {i}")

    # number = 1 means we only run it once in timeit 
    naive = timeit.timeit(
        lambda: [conflicts_naive(board) for board in boards], number=1
    )
    linear = timeit.timeit(
        lambda: [conflicts(board) for board in boards], number=1
    )
    jit = timeit.timeit(
        lambda: [conflicts_jit(board) for board in boards], number=1
    )

 
    x.append(i)
    naive_times.append(naive)
    linear_times.append(linear)
    jit_times.append(jit)

# average time per board in microseconds
naive_times = np.array(naive_times) / reps * 1e6
linear_times = np.array(linear_times) / reps * 1e6
jit_times = np.array(jit_times) / reps * 1e6

print("Times for 8x8 board in microseconds")
print(f'naive: {naive_times[0]}, linear: {linear_times[0]}, jit: {jit_times[0]}')
 
```

```python
plt.plot(x, naive_times)
plt.plot(x, linear_times)
plt.plot(x, jit_times)
plt.xlabel("board size")
plt.ylabel("runtime [microseconds - µs]")
plt.legend(["naive", "linear", "jit compiled with numba"])
plt.show()
```

```python
plt.plot(x, linear_times)
plt.plot(x, jit_times)
plt.xlabel("board size")
plt.ylabel("runtime [microseconds - µs]")
plt.legend(["linear", "jit compiled with numba"])
plt.show()
```

```python
# Bài 6: Vẽ đồ thị Log-Log để đối chiếu độ dốc (bậc độ phức tạp O(n^2) vs O(n)) và hệ số tăng tốc JIT
plt.figure(figsize=(9, 5))
plt.loglog(x, naive_times, 'o-', label="Naive O(n^2) (Python nested loops)")
plt.loglog(x, linear_times, 's-', label="Linear O(n) (Frequency array + comb2)")
plt.loglog(x, jit_times, '^-', label="Numba JIT O(n) (@njit compiled)")

plt.xlabel("Kích thước bàn cờ n (log scale)")
plt.ylabel("Thời gian chạy trung bình mỗi bàn cờ [µs] (log scale)")
plt.title("So sánh thời gian tính hàm conflicts() trên thang đo Log-Log")
plt.legend()
plt.grid(True, which="both", linestyle="--", alpha=0.5)
plt.show()

print(f"Tại n = {x[-1]}:")
print(f" - Tốc độ Linear O(n) nhanh hơn Naive O(n^2): {naive_times[-1] / linear_times[-1]:.1f} lần")
print(f" - Tốc độ Numba JIT O(n) nhanh hơn Linear Python O(n): {linear_times[-1] / jit_times[-1]:.1f} lần")
print(f" - Tốc độ Numba JIT O(n) nhanh hơn Naive O(n^2): {naive_times[-1] / jit_times[-1]:.1f} lần")
```

## Giải thích và Phân tích Tối ưu hóa với Numba (Bài 6)

1. **Cơ chế đếm xung đột $O(n)$ bằng mảng tần suất và công thức $C(k, 2)$:**
   * Thay vì dùng 2 vòng lặp lồng nhau duyệt qua mọi cặp cột $(i, j)$ mất thời gian $O(n^2)$ như `conflicts_naive()`, hàm `conflicts()` chỉ duyệt qua mảng `board` đúng 1 lần ($O(n)$) để đếm số lượng quân hậu nằm trên:
     * Cùng một hàng: chỉ số `board[i]` (có $n$ hàng).
     * Cùng đường chéo phụ ($\diagup$): tổng chỉ số hàng và cột không đổi `i + board[i]` (có $2n$ đường chéo).
     * Cùng đường chéo chính ($\diagdown$): hiệu chỉ số cột và hàng dịch chuyển `i - board[i] + n` (có $2n$ đường chéo).
   * Nếu trên một hàng hoặc một đường chéo có $k$ quân hậu, mọi cặp 2 quân hậu bất kỳ trong $k$ quân đó đều tấn công nhau (do biểu diễn trạng thái mỗi cột 1 quân nên không có quân nào chắn đúng vị trí trùng nhau). Do đó, số cặp xung đột sinh ra trên hàng/đường chéo đó chính là tổ hợp chập 2 của $k$:
     $$C(k, 2) = \binom{k}{2} = \frac{k(k - 1)}{2}$$
   * Tổng giá trị $C(k, 2)$ trên toàn bộ $n$ hàng và $4n$ đường chéo cho ta chính xác tổng số cặp xung đột $f(\text{board})$ chỉ với độ phức tạp thời gian và không gian $O(n)$.

2. **Vì sao phải gọi `conflicts_jit(random_board(4))` một lần trước khi đo thời gian?**
   * Numba hoạt động theo cơ chế **Just-In-Time (JIT) Compilation**: ở lần gọi đầu tiên, Numba phải phân tích kiểu dữ liệu (type inference) của đối số đầu vào và biên dịch mã bytecode Python sang mã máy gốc (native machine code qua LLVM). Quá trình biên dịch lần đầu này tốn khoảng vài chục đến vài trăm mili-giây. Nếu không gọi "warm-up" trước, thời gian biên dịch sẽ bị cộng dồn vào phép đo đầu tiên, làm sai lệch hoàn toàn kết quả đo thời gian thực thi thực tế (vốn chỉ tốn chưa tới $1\ \mu\text{s}$ trên bàn cờ $8 \times 8$).

3. **Khác biệt giữa giảm độ phức tạp thuật toán ($O(n^2) \to O(n)$) và biên dịch JIT:**
   * **Giảm độ phức tạp thuật toán ($O(n^2) \to O(n)$):** Thay đổi **độ dốc (bậc tăng trưởng tiệm cận)** của thời gian chạy khi $n$ lớn dần. Khi $n = 450$, thuật toán $O(n)$ thuần Python nhanh hơn thuật toán $O(n^2)$ hàng trăm lần và khoảng cách này càng giãn rộng khi $n \to \infty$.
   * **Biên dịch JIT (`@njit`):** Không làm thay đổi độ phức tạp tiệm cận (vẫn là $O(n)$, đường biểu diễn trên đồ thị log-log song song với đường Linear Python), nhưng loại bỏ chi phí thông dịch của Python (interpreter overhead, boxing/unboxing kiểu số nguyên) và tận dụng vector hóa CPU, giúp **giảm hệ số hằng số (constant factor)** thêm khoảng **10 - 15 lần** so với hàm $O(n)$ thuần Python.
